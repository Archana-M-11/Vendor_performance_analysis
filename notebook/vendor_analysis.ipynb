-- purchase_summary gives vendor wise purchse history --


```python
purchase_summary = purchases.groupby(
    ["VendorNumber", "VendorName"]
).agg({
    "Quantity": "sum",
    "PurchasePrice": "mean"
}).reset_index()

purchase_summary.head()
```

```python
sales_summary = sales.groupby(
    ["VendorNo", "VendorName"]
).agg({
    "SalesQuantity": "sum",
    "SalesDollars": "sum",
    "SalesPrice": "mean"
}).reset_index()

sales_summary.head()
```

```python
vendor_summary = purchase_summary.merge(
    sales_summary,
    left_on="VendorNumber",
    right_on="VendorNo",
    how="left"
)

vendor_summary.head()
```

```python
invoice_summary = vendor_invoice.groupby(
    ["VendorNumber", "VendorName"]
).agg({
    "Freight": "sum",
    "Dollars": "sum"
}).reset_index()

invoice_summary.head()
```

```python
vendor_summary = vendor_summary.merge(
    invoice_summary,
    on="VendorNumber",
    how="left"
)

vendor_summary.head()
```

```python
vendor_summary["EstimatedProfit"] = (
    vendor_summary["SalesDollars"]
    - vendor_summary["Dollars"]
    - vendor_summary["Freight"]
)

vendor_summary.head()
```

```python
vendor_summary[[
    "VendorName",
    "SalesDollars",
    "Dollars",
    "Freight",
    "EstimatedProfit"
]].head(10)
```

```python
final_vendor_analysis = vendor_summary[[
    "VendorNumber",
    "VendorName",
    "SalesQuantity",
    "SalesDollars",
    "Quantity",
    "Dollars",
    "Freight",
    "EstimatedProfit"
]]

final_vendor_analysis.head()
```

```python
final_vendor_analysis.to_csv(
    "final_vendor_analysis.csv",
    index=False
)
```

```python
brand_summary = sales.groupby("Description").agg({
    "SalesDollars": "sum"
}).reset_index()

brand_summary.head()
```

```python
brand_summary.to_csv(
    "brand_summary.csv",
    index=False
)
```

```python
brand_summary = sales.groupby("Description").agg({
    "SalesDollars": "sum",
    "SalesPrice": "mean"
}).reset_index()

brand_summary["ProfitMargin"] = (
    brand_summary["SalesPrice"] / 
    brand_summary["SalesPrice"].max()
) * 100

brand_summary.head()
```

```python
brand_summary.to_csv(
    "brand_summary.csv",
    index=False
)
```
